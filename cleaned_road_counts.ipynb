```python
columns = [
    "count_point_id", "year", "road_name",
    "easting", "northing", "road_type",
    "estimation_method", "all_motor_vehicles"
]
data = data[columns]
```

```python
print(data["estimation_method"].value_counts())
```

```python
# Calculate percentages
print(data["estimation_method"].value_counts(normalize=True) * 100)

# Compare Major and Minor roads
print(pd.crosstab(data["road_type"], data["estimation_method"]))
```

```python
# Save cleaned dataset with estimation_method
data.to_csv("dft_aadf_region_id_6.csv", index=False)

# Save a summary for your cleaning log
split = data["estimation_method"].value_counts()
split.to_csv("estimation_method_summary.csv")
```

```python
print("Vehicle counts of zero or negative:", (data["all_motor_vehicles"] <= 0).sum())
```

```python
data = data[(data["all_motor_vehicles"] > 0)].copy()
```

```python
data.shape
```

```python
print(sorted(data["year"].unique()))
```

```python
# Average including 2021
avg_with_2021 = data["all_motor_vehicles"].mean()

# Average excluding 2021
avg_without_2021 = data.loc[
    data["year"] != 2021, "all_motor_vehicles"
].mean()

# Percentage difference
pct_change = (
    (avg_without_2021 - avg_with_2021)
    / avg_with_2021
) * 100

print("With 2021:", round(avg_with_2021, 2))
print("Without 2021:", round(avg_without_2021, 2))
print("Percentage change:", round(pct_change, 2), "%")
```
